# Turnaround Activity Analysis

Analyze turnaround activity patterns to optimize duration estimates and identify bottlenecks.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta
```

## Activity Duration Analysis

```python
# Sample activity data
activities = pd.DataFrame({
    'activity_type': ['cleaning', 'refueling', 'catering', 'boarding'] * 250,
    'duration_minutes': np.random.randint(10, 40, 1000),
    'aircraft_type': np.random.choice(['A321', 'B737', 'A330'], 1000)
})

# Average duration by activity type
avg_durations = activities.groupby('activity_type')['duration_minutes'].agg(['mean', 'std', 'min', 'max'])
print('Average Activity Durations:')
print(avg_durations)
```

## Critical Path Analysis

```python
# Visualize activity durations
plt.figure(figsize=(12, 6))
activities.boxplot(column='duration_minutes', by='activity_type', figsize=(12, 6))
plt.xlabel('Activity Type')
plt.ylabel('Duration (minutes)')
plt.title('Activity Duration Distribution')
plt.suptitle('')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Bottleneck Identification

```python
# Identify activities that frequently cause delays
activities['is_delayed'] = activities['duration_minutes'] > activities.groupby('activity_type')['duration_minutes'].transform('median') * 1.2

delay_frequency = activities.groupby('activity_type')['is_delayed'].sum() / activities.groupby('activity_type')['is_delayed'].count() * 100

plt.figure(figsize=(10, 6))
delay_frequency.sort_values(ascending=True).plot(kind='barh')
plt.xlabel('Delay Frequency (%)')
plt.title('Activity Delay Frequency')
plt.tight_layout()
plt.show()
```

## Recommendations

Based on the analysis:
1. Focus on optimizing high-delay-frequency activities
2. Monitor critical path activities closely
3. Implement buffers for activities with high variance
4. Consider parallel execution where possible
