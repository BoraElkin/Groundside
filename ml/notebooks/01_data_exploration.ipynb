# Ground Operations Data Exploration

This notebook explores historical flight turnaround data to identify patterns and features for delay prediction.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta

sns.set_style('whitegrid')
%matplotlib inline
```

## Load Data

Load historical turnaround and delay data from the database or CSV files.

```python
# Example: Load from CSV (replace with actual database connection)
# df = pd.read_csv('../data/historical_turnarounds.csv')

# For demo, create sample data
np.random.seed(42)
n_samples = 1000

df = pd.DataFrame({
    'flight_number': [f'TK{i:04d}' for i in range(n_samples)],
    'aircraft_type': np.random.choice(['A321', 'B737', 'A330', 'B777'], n_samples),
    'passenger_count': np.random.randint(50, 300, n_samples),
    'scheduled_duration': np.random.randint(30, 90, n_samples),
    'actual_duration': np.random.randint(30, 120, n_samples),
    'delay_minutes': np.random.randint(-5, 60, n_samples),
    'hour_of_day': np.random.randint(0, 24, n_samples),
    'gate_congestion': np.random.uniform(0, 1, n_samples)
})

df.head()
```

## Data Overview

```python
df.info()
df.describe()
```

## Delay Distribution

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(df['delay_minutes'], bins=30, edgecolor='black')
plt.xlabel('Delay (minutes)')
plt.ylabel('Frequency')
plt.title('Distribution of Turnaround Delays')

plt.subplot(1, 2, 2)
df.boxplot(column='delay_minutes', by='aircraft_type', figsize=(10, 5))
plt.xlabel('Aircraft Type')
plt.ylabel('Delay (minutes)')
plt.title('Delays by Aircraft Type')
plt.suptitle('')

plt.tight_layout()
plt.show()
```

## Correlation Analysis

```python
# Correlation matrix
numeric_cols = ['passenger_count', 'scheduled_duration', 'actual_duration', 'delay_minutes', 'hour_of_day', 'gate_congestion']
corr_matrix = df[numeric_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Feature Correlation Matrix')
plt.show()
```

## Time-based Patterns

```python
# Average delay by hour of day
hourly_delays = df.groupby('hour_of_day')['delay_minutes'].mean()

plt.figure(figsize=(12, 5))
plt.bar(hourly_delays.index, hourly_delays.values)
plt.xlabel('Hour of Day')
plt.ylabel('Average Delay (minutes)')
plt.title('Average Delay by Hour of Day')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

## Key Findings

1. Delay distribution and characteristics
2. Impact of aircraft type on delays
3. Peak delay hours
4. Correlation between features and delays

These insights will inform feature engineering for the delay prediction model.
